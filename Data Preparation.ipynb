```python
import pandas as pd

# Read the CSV into a DataFrame
df = pd.read_csv('pp-complete.csv', header=None)

# Display the result
df.head()
```

```python
# List of official column names for the Price Paid dataset
df.columns = [
    'TID', 'Price', 'Date', 'Postcode', 'Type', 'New_Build', 
    'Tenure', 'PAON', 'SAON', 'Street', 'Locality', 
    'Town', 'District', 'County', 'Category', 'Status'
]

# Verify the changes
df.head()
```

```python
# Filter the DataFrame for London
london_df = df[df['Town'] == 'LONDON']

# Display the first few rows of the filtered data
london_df.head()

# Optional: Check how many records were found
print(f"Total records found in London: {len(london_df)}")
```
